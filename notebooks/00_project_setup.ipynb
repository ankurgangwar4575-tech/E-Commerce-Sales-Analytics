```python
from pathlib import Path
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

RANDOM_SEED=42
random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)

pd.set_option("display.max_columns",100)
pd.set_option("display.float_format","{:,.2f}".format)
sns.set_theme(style="whitegrid")

print("Setup imports completed successfully")
```

# Project paths

```python
CURRENT_DIR = Path.cwd().resolve()

if CURRENT_DIR.name == "notebooks":
    PROJECT_ROOT = CURRENT_DIR.parent
else:
    PROJECT_ROOT = CURRENT_DIR

DATA_DIR = PROJECT_ROOT / "data"
NOTEBOOKS_DIR = PROJECT_ROOT / "notebooks"
SRC_DIR = PROJECT_ROOT / "src"
MODELS_DIR = PROJECT_ROOT / "models"
TESTS_DIR = PROJECT_ROOT / "tests"

print(f"Project root: {PROJECT_ROOT}")
print(f"Data directory: {DATA_DIR}")
```

# Required source files

```python
DATA_FILES = {
    "customers": DATA_DIR / "customer_master.csv",
    "orders": DATA_DIR / "ecommerce_sales_customer_analytics_150k.csv",
    "order_items": DATA_DIR / "order_items.csv",
    "products": DATA_DIR / "product_catalog.csv",
    "statistics": DATA_DIR / "dataset_statistics.csv",
}

missing_files = [name for name, path in DATA_FILES.items() if not path.exists()]

if missing_files:
    raise FileNotFoundError(
        f"Missing required data files: {', '.join(missing_files)}"
    )

for name, path in DATA_FILES.items():
    file_size_mb = path.stat().st_size / (1024 ** 2)
    print(f"{name:<12} | {path.name:<50} | {file_size_mb:>7.2f} MB")
```

# Load source data

```python
customers_df = pd.read_csv(DATA_FILES["customers"])
orders_df = pd.read_csv(DATA_FILES["orders"])
order_items_df = pd.read_csv(DATA_FILES["order_items"])
products_df = pd.read_csv(DATA_FILES["products"])
statistics_df = pd.read_csv(DATA_FILES["statistics"])

datasets = {
    "Customers": customers_df,
    "Orders": orders_df,
    "Order items": order_items_df,
    "Products": products_df,
    "Dataset statistics": statistics_df,
}

for name, dataframe in datasets.items():
    print(f"{name:<20} Rows: {dataframe.shape[0]:>8,} | Columns: {dataframe.shape[1]:>3,}")
```
